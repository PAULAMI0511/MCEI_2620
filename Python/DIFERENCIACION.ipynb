```python
import numpy as np
import matplotlib.pyplot as plt
```

cargar datos 

```python
data = np.loadtxt("trayectoria_robot.csv", delimiter=",", skiprows=1)
```

```python
t,x,y = data[:,0], data[:,1], data[:,2]
vx = np.gradient(x, t)
vy = np.gradient(y, t)
v = np.sqrt(vx**2 + vy**2)
theta = np.unwrap(np.arctan2(vy, vx))
omega = np.gradient(theta, t)
```

# Imprimir tabla de primeros 5 datos calculados

```python

print(f"{'t [s]':>6} | {'x [m]':>8} | {'y [m]':>8} | {'v [m/s]':>8} | {'theta [rad]':>11} | {'omega [rad/s]':>13}")
print("-" * 65)
for i in range(5):
    print(f"{t[i]:6.2f} | {x[i]:8.4f} | {y[i]:8.4f} | {v[i]:8.4f} | {theta[i]:11.4f} | {omega[i]:13.4f}")
```

CREACION DE GRAFICAS - cuadricula 

```python
fig, axs = plt.subplots(2, 2, figsize=(11, 8))

axs[0, 0].plot(x, y, "b.-")
axs[0, 0].set_title("Trayectoria del Robot (y vs x)")
axs[0, 0].set_xlabel("x [m]")
axs[0, 0].set_ylabel("y [m]")
axs[0, 0].grid(True)

axs[0, 1].plot(t, v, "r.-")
axs[0, 1].set_title("Velocidad Lineal (v vs t)")
axs[0, 1].set_xlabel("t [s]")
axs[0, 1].set_ylabel("v [m/s]")
axs[0, 1].grid(True)

axs[1, 0].plot(t, theta, "g.-")
axs[1, 0].set_title("Orientación (θ vs t)")
axs[1, 0].set_xlabel("t [s]")
axs[1, 0].set_ylabel("θ [rad]")
axs[1, 0].grid(True)

axs[1, 1].plot(t, omega, "m.-")
axs[1, 1].set_title("Velocidad Angular (ω vs t)")
axs[1, 1].set_xlabel("t [s]")
axs[1, 1].set_ylabel("ω [rad/s]")
axs[1, 1].grid(True)

plt.tight_layout()
plt.show()
```
