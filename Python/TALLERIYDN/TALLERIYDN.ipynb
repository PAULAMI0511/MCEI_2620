```python
import numpy as np
import time
from scipy.integrate import quad

def f(x):
    return np.exp(-0.4 * x) * (1.0 + 0.5 * np.sin(3.0 * x))

def trapecio_compuesto(f, a, b, n):
    x = np.linspace(a, b, n + 1)
    y = f(x)
    h = (b - a) / n
    return h * (0.5 * y[0] + np.sum(y[1:-1]) + 0.5 * y[-1])

a = 0.0
b = 8.0
I_ref = 2.30231617823528
N_vals = [10, 20, 50, 100, 500, 1000]
```

BUCLE INTEGRAL 

```python
print("=========================================================================")
print("  n   |  Integral Aproximada (Trap) |   Error Absoluto   |   Tiempo (ms)")
print("=========================================================================")

for n in N_vals:
    start = time.perf_counter()
    I_trap = trapecio_compuesto(f, a, b, n)
    end = time.perf_counter()
    
    t_exec = (end - start) * 1000.0
    err = abs(I_trap - I_ref)
    
    # El print DEBE ir indentado dentro del for
    print(f"{n:<5d} | {I_trap:<27.8f} | {err:<18.2e} | {t_exec:<12.6f}")

print("=========================================================================")
```

EVALUACION DE LA INTEGRAL CON EL MÉTODO 

```python
start_quad = time.perf_counter()
I_quad, err_est = quad(f, a, b)
end_quad = time.perf_counter()

t_quad = (end_quad - start_quad) * 1000.0

print("--- SEGUNDA APROXIMACIÓN: SciPy (quad) ---")
print(f"Integral aproximada (SciPy) : {I_quad:.8f}")
print(f"Error estimado SciPy        : {err_est:.2e}")
print(f"Error real vs I_ref         : {abs(I_quad - I_ref):.2e}")
print(f"Tiempo de ejecución         : {t_quad:.6f} ms")
print("=========================================================================")
```
