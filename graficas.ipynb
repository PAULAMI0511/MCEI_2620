```python
import numpy as np
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
```

```python
puntos = np.loadtxt("puntos.txt")
hull = np.loadtxt("hull.txt")
aristas = np.loadtxt("aristas.txt")

fig, ax = plt.subplots(figsize=(7, 7))

# 1. Aristas de Delaunay (van debajo, en gris)
for x1, y1, x2, y2 in aristas:
    ax.plot([x1, x2], [y1, y2], color="gray", linewidth=0.6)

# 2. Envolvente convexa (en rojo, cerrando el poligono)
hull_cerrado = np.vstack([hull, hull[0]])
ax.plot(hull_cerrado[:, 0], hull_cerrado[:, 1], color="red", linewidth=2)

# 3. Los sensores (encima, en negro)
ax.scatter(puntos[:, 0], puntos[:, 1], s=15, color="black")

ax.set_aspect("equal")
ax.set_title("Sensores, envolvente convexa y triangulacion de Delaunay")
plt.savefig("figura.png", dpi=150)
```
